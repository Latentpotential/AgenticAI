```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-120b")
```

```python
model.invoke("Please tell what is AI Intelligence in simple words and give an example of it.")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert."),
    HumanMessage("Write a poem about AI."), 
]

response = model.invoke(messages)
print(response.content)
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages = [
    system_msg,
    HumanMessage("Write a python function for fibonacci series.")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("You are a Senior Python coder.")

messages = [
    system_msg,
    HumanMessage("Write a python function for fibonacci series.")
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

ai_msg = AIMessage("Hello, I am an AI language model. How can I assist you today?")

messages = [
    SystemMessage("You are a helpful coding assistant."),
    HumanMessage("Write a python function for fibonacci series."),
    ai_msg,
    HumanMessage("Can you explain how the function works?"),
]

response = model.invoke(messages)
print(response.content)
```

```python


from langchain.messages import ToolMessage


ai_message = AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args" : {"location": "New York"},
        "id": "call_1"
    }]
)

weather_result = "Sunny, 73 F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_1"
)

messages = [
    HumanMessage("What is the weather like in New York?"),
    ai_message,
    tool_message
]

response = model.invoke(messages)
print(response.content)
```
